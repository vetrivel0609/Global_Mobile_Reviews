```python
#Import Libraries

import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

```python
#Set Paths

BASE_DIR = os.path.dirname(os.getcwd())
INPUT_PATH = os.path.join(BASE_DIR, "outputs", "cleaned_mobile_reviews.csv")
OUTPUT_PATH = os.path.join(BASE_DIR, "outputs", "clustered_mobile_reviews.csv")

print("Input:", INPUT_PATH)
print("Output:", OUTPUT_PATH)
```

```python
#Load Cleaned Data

df = pd.read_csv(INPUT_PATH)
print("Shape:", df.shape)
df.head()
```

```python
#Check Required Features

cluster_features = [
    "price_usd",
    "rating",
    "battery_life_rating",
    "camera_rating",
    "performance_rating",
    "design_rating",
    "display_rating"
]
df[cluster_features].describe().T
```

```python
#Create Clustering Dataset

X = df[cluster_features].copy()

print("Clustering data shape:", X.shape)
print("Missing values:", X.isna().sum().sum())
```

```python
#Scale Features

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Scaled data shape:", X_scaled.shape)
```
